```python

# Cell 3: Combined Two-Tier Action Engine (Fast Sieve + Deep VLM)
import os
import re
import cv2
import json
import time
import requests
import numpy as np
from PIL import Image
from fastapi import FastAPI, Request
from fastapi.responses import StreamingResponse
from qwen_vl_utils import process_vision_info
from scenedetect import detect, ContentDetector

app = FastAPI(title="Two-Tier CineCutter A100 Engine")

BATCH_SIZE = 16

ACTION_SYSTEM_PROMPT = (
    "You are an elite Hollywood action cinema editor. Analyze these sequence frames with zero tolerance for non-action.\n\n"
    "OBJECTIVE:\n"
    "Determine if this scene depicts genuine, high-stakes CINEMATIC ACTION.\n\n"
    "INCLUSION CRITERIA (Mark is_action: true):\n"
    "1. Hand-to-Hand & Melee Combat: Martial arts strikes, grapples, takedowns, throws, knife/sword fighting, blunt weapon impacts, brutal choreography.\n"
    "2. Gunplay & Tactical Firefights: Active discharges, CQB room clearing, muzzle flashes, cover-to-cover shooting, sniper engagements, armed standoffs with lethal tension.\n"
    "3. Vehicular Pursuits & Destruction: High-speed car/motorcycle chases, vehicle collisions, ramming, dynamic drifting under fire, bailouts.\n"
    "4. Explosions & Environmental Chaos: Detonations, heavy structural collapse, grenade blasts, rocket/artillery fire, characters diving from blasts.\n"
    "5. High-Stakes Physical Stunts: Parkour escapes under hot pursuit, combat falls, scaling extreme heights, train/helicopter roof sequences.\n\n"
    "STRICT EXCLUSION CRITERIA (Mark is_action: false):\n"
    "- Heated arguments, shouting, or emotional drama without physical strikes.\n"
    "- Aiming a weapon casually without impending combat or discharge.\n"
    "- Casual walking, jogging, or running without active tactical pursuit/threat.\n"
    "- Sports, athletics, gymnastics, dancing, or comedic slapstick.\n"
    "- Static conversations, walking down corridors, vehicle cruising/talking inside cars.\n\n"
    "Output strictly valid JSON:\n"
    "{\n"
    '  "is_action": true,\n'
    '  "action_type": "hand-to-hand | gunfight | vehicle chase | explosion | tactical raid | melee | none",\n'
    '  "confidence": 0.0 to 1.0,\n'
    '  "reasoning": "Brief factual description of the combat, kinetic threat, or choreography observed"\n'
    "}"
)

def timecode(seconds):
    h = int(seconds // 3600)
    m = int((seconds % 3600) // 60)
    s = seconds % 60
    return f"{h:02d}:{m:02d}:{s:06.3f}"

def extract_clip_frames(cap, start_sec, end_sec, fps):
    """Extracts 2 frames for fast OpenCLIP screening."""
    duration = end_sec - start_sec
    mid_sec = start_sec + (duration * 0.5)
    indices = [int(start_sec * fps), int(mid_sec * fps)]
    tensors = []
    for idx in indices:
        cap.set(cv2.CAP_PROP_POS_FRAMES, idx)
        ret, frame = cap.read()
        if ret:
            rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            pil_img = Image.fromarray(rgb)
            tensors.append(clip_preprocess(pil_img).unsqueeze(0))
    return tensors

def extract_qwen_frames(cap, start_sec, end_sec, fps):
    """Zero-disk in-memory extraction (8-12 frames) for deep VLM reasoning."""
    duration = end_sec - start_sec
    num_frames = 8 if duration < 3.0 else 12
    start_frame = int(start_sec * fps)
    end_frame = int(end_sec * fps)
    if end_frame <= start_frame:
        end_frame = start_frame + int(fps)

    indices = np.linspace(start_frame, end_frame - 1, num=num_frames, dtype=int)
    pil_frames = []
    for idx in indices:
        cap.set(cv2.CAP_PROP_POS_FRAMES, idx)
        ret, frame = cap.read()
        if ret:
            h, w = frame.shape[:2]
            scale = 512 / max(h, w)
            if scale < 1.0:
                frame = cv2.resize(frame, (int(w * scale), int(h * scale)), interpolation=cv2.INTER_AREA)
            rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            pil_frames.append(Image.fromarray(rgb))
    return pil_frames

def evaluate_qwen_batch(batch_scenes_data):
    """Batched forward pass through Qwen2.5-VL."""
    batch_messages = []
    for item in batch_scenes_data:
        msg = [
            {
                "role": "user",
                "content": [
                    {"type": "video", "video": item["frames"], "fps": 2.0},
                    {"type": "text", "text": ACTION_SYSTEM_PROMPT}
                ]
            }
        ]
        batch_messages.append(msg)

    texts = [qwen_processor.apply_chat_template(m, tokenize=False, add_generation_prompt=True) for m in batch_messages]
    all_image_inputs = []
    all_video_inputs = []
    for m in batch_messages:
        img_inp, vid_inp = process_vision_info(m)
        all_image_inputs.extend(img_inp if img_inp else [])
        all_video_inputs.extend(vid_inp if vid_inp else [])

    inputs = qwen_processor(
        text=texts,
        images=all_image_inputs if all_image_inputs else None,
        videos=all_video_inputs if all_video_inputs else None,
        padding=True,
        return_tensors="pt"
    ).to("cuda")

    with torch.no_grad():
        generated_ids = qwen_model.generate(**inputs, max_new_tokens=96, do_sample=False)
        generated_ids_trimmed = [
            out_ids[len(in_ids):] for in_ids, out_ids in zip(inputs.input_ids, generated_ids)
        ]
        decoded = qwen_processor.batch_decode(
            generated_ids_trimmed, skip_special_tokens=True, clean_up_tokenization_spaces=False
        )

    results = []
    for raw in decoded:
        try:
            match = re.search(r"\{.*?\}", raw, re.DOTALL)
            results.append(json.loads(match.group(0)) if match else json.loads(raw))
        except Exception:
            results.append({"is_action": False, "action_type": "parse_err", "confidence": 0.0})
    return results

def stream_action_analysis(video_source, threshold=0.60):
    yield json.dumps({"type": "status", "msg": "[*] Connected to Dual-Engine (OpenCLIP + Qwen2.5-VL A100)..."}) + "\n"

    local_tmp_video = "/tmp/current_movie.mp4"
    if os.path.exists(local_tmp_video):
        os.remove(local_tmp_video)

    yield json.dumps({"type": "status", "msg": "[*] Streaming movie from VPS Media Bridge..."}) + "\n"
    with requests.get(video_source, stream=True, timeout=60) as r:
        r.raise_for_status()
        with open(local_tmp_video, "wb") as f:
            for chunk in r.iter_content(chunk_size=1024 * 1024 * 8):
                if chunk:
                    f.write(chunk)

    yield json.dumps({"type": "status", "msg": "[✓] Ingestion complete. Running PySceneDetect..."}) + "\n"
    scene_list = detect(local_tmp_video, ContentDetector(threshold=28.0, min_scene_len=30))
    total_scenes = len(scene_list)
    yield json.dumps({"type": "status", "msg": f"[✓] Found {total_scenes} takes. Initiating Stage 1: OpenCLIP Fast Sieve..."}) + "\n"

    cap = cv2.VideoCapture(local_tmp_video)
    fps = cap.get(cv2.CAP_PROP_FPS) or 24.0

    # -------------------------------------------------------------
    # STAGE 1: High-Recall OpenCLIP Sieve (Runs in ~60-90 seconds)
    # -------------------------------------------------------------
    candidates = []
    CLIP_SIEVE_THRESHOLD = 0.32  # Loose threshold to retain subtle action

    for idx, scene in enumerate(scene_list, 1):
        start_sec = scene[0].seconds
        end_sec = scene[1].seconds
        duration = round(end_sec - start_sec, 2)
        if duration < 1.0:
            continue

        clip_tensors = extract_clip_frames(cap, start_sec, end_sec, fps)
        if not clip_tensors:
            continue

        batch_tensor = torch.cat(clip_tensors).to(device=DEVICE, dtype=torch.float16)
        with torch.no_grad():
            img_feats = clip_model.encode_image(batch_tensor)
            img_feats /= img_feats.norm(dim=-1, keepdim=True)
            sim = (100.0 * img_feats @ clip_text_features.T).softmax(dim=-1)

        # Average action confidence across sampled frames
        mean_probs = sim.mean(dim=0).cpu().numpy()
        action_score = float(mean_probs[clip_action_indices].sum())

        if action_score >= CLIP_SIEVE_THRESHOLD:
            candidates.append({
                "idx": idx,
                "start_sec": start_sec,
                "end_sec": end_sec,
                "duration": duration,
                "clip_score": round(action_score, 3)
            })

        if idx % 150 == 0 or idx == total_scenes:
            yield json.dumps({
                "type": "progress",
                "msg": f"[Stage 1] OpenCLIP screened {idx}/{total_scenes} takes... (Passed {len(candidates)} candidates)"
            }) + "\n"

    sieve_rejection_pct = round(100 - (len(candidates) / max(total_scenes, 1) * 100), 1)
    yield json.dumps({
        "type": "status",
        "msg": f"[✓] Stage 1 Finished: Discarded {sieve_rejection_pct}% non-action cuts. Commencing Stage 2: Qwen2.5-VL Deep Audit on {len(candidates)} candidate scenes..."
    }) + "\n"

    # -------------------------------------------------------------
    # STAGE 2: Deep VLM Reasoning on Candidates Only
    # -------------------------------------------------------------
    actions = []
    batch_queue = []

    for c_idx, item in enumerate(candidates, 1):
        frames = extract_qwen_frames(cap, item["start_sec"], item["end_sec"], fps)
        if not frames:
            continue

        batch_queue.append({
            "idx": item["idx"],
            "start_sec": item["start_sec"],
            "end_sec": item["end_sec"],
            "duration": item["duration"],
            "frames": frames
        })

        if len(batch_queue) == BATCH_SIZE or c_idx == len(candidates):
            batch_results = evaluate_qwen_batch(batch_queue)

            for cand, res in zip(batch_queue, batch_results):
                is_act = res.get("is_action", False)
                conf = float(res.get("confidence", 0.0))
                act_type = res.get("action_type", "action")
                reason = res.get("reasoning", "")

                if is_act and conf >= threshold:
                    entry = {
                        "start_seconds": cand["start_sec"],
                        "end_seconds": cand["end_sec"],
                        "start_timecode": timecode(cand["start_sec"]),
                        "end_timecode": timecode(cand["end_sec"]),
                        "duration_seconds": cand["duration"],
                        "action_type": act_type,
                        "confidence": conf,
                        "reasoning": reason
                    }
                    actions.append(entry)
                    time_label = f"{timecode(cand['start_sec'])} - {timecode(cand['end_sec'])}"
                    yield json.dumps({
                        "type": "action_hit",
                        "progress": f"{cand['idx']}/{total_scenes}",
                        "scene": time_label,
                        "label": f"{act_type} — {reason}",
                        "confidence": conf
                    }) + "\n"

            yield json.dumps({
                "type": "progress",
                "msg": f"[Stage 2] VLM audited {c_idx}/{len(candidates)} candidates... ({len(actions)} verified sequences)"
            }) + "\n"

            batch_queue = []

    cap.release()
    if os.path.exists(local_tmp_video):
        os.remove(local_tmp_video)

    yield json.dumps({"type": "complete", "actions": actions}) + "\n"

@app.get("/health")
def health():
    return {
        "status": "healthy",
        "gpu": torch.cuda.get_device_name(0) if torch.cuda.is_available() else "None",
        "vram_allocated_gb": round(torch.cuda.memory_allocated() / (1024**3), 2),
        "engine": "Two-Tier OpenCLIP ViT-L-14 + Qwen2.5-VL-7B (A100)"
    }

@app.post("/analyze_stream")
async def analyze_stream_endpoint(request: Request):
    payload = await request.json()
    video_source = payload.get("video_source")
    threshold = float(payload.get("threshold", 0.60))
    return StreamingResponse(
        stream_action_analysis(video_source, threshold),
        media_type="application/x-ndjson"
    )
```
